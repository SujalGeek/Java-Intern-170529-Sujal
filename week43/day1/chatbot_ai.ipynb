Building the chatbot with the langchain

```python
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
```

```python
os.environ['GROQ_API_KEY'] = ""
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.1
)
```

```python
parser = StrOutputParser()
```

chat_history = list of messages
mesages = role and content
role: 1. system; 1. user; 2. assistant;

```python
def chat():
  chat_history = [
      ("system","You are a helpful chatbot. Be concise and accurate")
  ]

  print("LangChain chatbot: Type 'exit' to quit")

  while True:
    user_input = input("You: ").strip()

    if user_input.lower() == "exit":
      break

    # add user message
    chat_history.append(("user",user_input))

    # build the prompt dynamically
    prompt = ChatPromptTemplate.from_messages(chat_history)

    chain = prompt | llm | parser

    response = chain.invoke({})

    print(f"Bot: {response}\n")

    chat_history.append(("assistant",response))

    print("-"*80)

```

```python
chat()
```

```python
import os
from llama_index.core.llms import ChatMessage,MessageRole
from llama_index.llms.groq import Groq
```

```python
os.environ["GROQ_API_KEY"] = ""
```

```python
llm = Groq(
    model="llama-3.3-70b-versatile",
    temperature=0.1
)
```

```python
def chat():
  history = [
      ChatMessage(
          role= MessageRole.SYSTEM,
          content= "You are helpful assistant. Be concise and accurate."
      )
  ]


  print("LlamaIndex ChatBot: Type 'exit' or quit.\n")

  while True:
    user = input("You: ").strip()
    if user.lower() in {"exit","quit"}:
      break
    if not user:
      continue


    history.append(ChatMessage(role=MessageRole.USER,content=user))

    resp = llm.chat(messages=history)

    answer = resp.message.content

    print(f"Bot: {answer}\n")

    history.append(ChatMessage(role=MessageRole.ASSISTANT,content=answer))

    print("-"*80)
```

```python
chat()
```
