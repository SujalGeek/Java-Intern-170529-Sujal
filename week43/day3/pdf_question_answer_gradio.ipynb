```python
%%writefile rag_utility.py
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import PyPDFLoader
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

working_dir = os.path.dirname(os.path.abspath(__file__))
embedding = HuggingFaceEmbeddings()
llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0)

def process_document_to_chroma_db(file_path):
    # Gradio passes the temp file path directly
    loader = PyPDFLoader(file_path)
    documents = loader.load()

    text_splitter = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap=200)
    texts = text_splitter.split_documents(documents)

    Chroma.from_documents(
        documents=texts,
        embedding=embedding,
        persist_directory=f"{working_dir}/doc_vectorstore"
    )
    return "✅ Document Processed Successfully"

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

def answer_question(user_question):
    # THE CHROMADB FIX: Using 'embedding_function' for loading
    vectordb = Chroma(
        persist_directory=f"{working_dir}/doc_vectorstore",
        embedding_function=embedding
    )
    retriever = vectordb.as_retriever()

    prompt = PromptTemplate.from_template(
        "Use the following pieces of retrieved context to answer the question.\n\nContext: {context}\n\nQuestion: {question}\n\nAnswer:"
    )

    rag_chain = (
        {"context": retriever | format_docs, "question": RunnablePassthrough()}
        | prompt
        | llm
        | StrOutputParser()
    )

    return rag_chain.invoke(user_question)
```

```python
%%writefile app.py
import gradio as gr
from rag_utility import process_document_to_chroma_db, answer_question

def process_pdf(pdf_file):
    if pdf_file is None:
        return "⚠️ Please upload a PDF first."
    # Gradio files have a .name attribute containing the file path
    return process_document_to_chroma_db(pdf_file.name)

def answer_query(question):
    if not question or not question.strip():
        return "⚠️ Please enter a valid question."
    return answer_question(question)

# The Gradio UI Layout
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🦙 Llama-3.3-70b Document RAG")

    with gr.Row():
        with gr.Column():
            pdf_input = gr.File(label="Upload a PDF File", file_types=[".pdf"])
            process_btn = gr.Button("Process Document", variant="primary")
            status_output = gr.Textbox(label="Status", interactive=False)

        with gr.Column():
            question_input = gr.Textbox(label="Ask your question about the document", lines=3)
            answer_btn = gr.Button("Answer", variant="primary")
            answer_output = gr.Textbox(label="Llama-3.3-70B Response", lines=10, interactive=False)

    # Connect buttons to the functions
    process_btn.click(fn=process_pdf, inputs=pdf_input, outputs=status_output)
    answer_btn.click(fn=answer_query, inputs=question_input, outputs=answer_output)

if __name__ == "__main__":
    print("Generating secure public link...")
    demo.launch(share=True, debug=True)
```

```python
import os
import subprocess
import time
import re

print("Cleaning up old zombie processes...")
# 1. The Kill Switch: Force quit old tunnels and streamlit instances
os.system("pkill -f streamlit")
os.system("pkill -f cloudflared")
os.system("rm -f /tmp/cloudflared")
time.sleep(2) # Give Linux a second to clear the memory locks

print("Downloading fresh Cloudflare Tunnel...")
# 2. Download fresh
os.system("wget -q -O /tmp/cloudflared https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64")
os.system("chmod +x /tmp/cloudflared")

print("Starting Streamlit Server...")
# 3. Start App
subprocess.Popen(["streamlit", "run", "app.py", "--server.enableCORS=false", "--server.enableXsrfProtection=false"])
time.sleep(3)

print("Opening Cloudflare Tunnel... (Give it 5-10 seconds)")
# 4. Start Tunnel
p = subprocess.Popen(["/tmp/cloudflared", "tunnel", "--url", "http://localhost:8501"], stderr=subprocess.PIPE, text=True)

# 5. Extract URL
for line in p.stderr:
    url_match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
    if url_match:
        print(f"\n=========================================================")
        print(f"✅ CLICK THIS EXACT URL: {url_match.group(0)}")
        print(f"=========================================================\n")
        break
```
